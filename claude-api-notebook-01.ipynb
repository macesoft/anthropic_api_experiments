```python
# Get the secret variable stored in the .env file

from dotenv import load_dotenv

load_dotenv()
```

```python
# Create an API client using the API key from the environment variable
from anthropic import Anthropic

anthropic_client = Anthropic()
```

```python
class ContextChatter:
    def __init__(self, client, max_tokens=1000, model="claude-haiku-4-5-20251001"):
        self.__client = client
        self.__model = model
        self.__max_tokens = max_tokens
        self.__messages = []

    def _store_user_message(self, text: str):
        self.__messages.append({"role": "user", "content": text})

    def _store_assistant_message(self, text: str):
        self.__messages.append({"role": "assistant", "content": text})

    def _send_to_claude(self):
        response = self.__client.messages.create(
            model=self.__model,
            max_tokens=self.__max_tokens,
            messages=self.__messages,
        )
        assistant_text = response.content[0].text
        return assistant_text
    
    def show_all_messages(self):
        for message in self.__messages:
            role = message["role"]
            content = message["content"]
            print(f"{role.capitalize()}: {content}")

    def init_chat(self):
        while True:
            user_input = input("You: ")
            
            if not user_input.strip():  
                print("Please enter a message.")
                continue

            if user_input.lower() in ["exit", "quit"]:
                print("Exiting chat...")
                break
            
            self._store_user_message(user_input)
            print(f"[You   ]: {user_input}") 

            assistant_response = self._send_to_claude()
            self._store_assistant_message(assistant_response)
            print(f"[Claude]: {assistant_response}")
            
"""
            show_all_messages = input("Do you want to see all messages? (y/n): ")
            if show_all_messages.lower() == "y":
                self.show_all_messages()
"""
```

```python
chatter = ContextChatter(anthropic_client)
chatter.init_chat()
```
